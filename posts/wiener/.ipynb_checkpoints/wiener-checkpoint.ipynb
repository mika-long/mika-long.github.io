First we look at the "standard" Brownian motion, which "involves Brownian motion paths with zero mean and unit variance". The following recursion relation is for a Brownian motion path across $n$ points: $$W(t_{i+1}) = W(t_i) + \sqrt{t_{i+1}-t_i} Z_{i+1}$$ where $Z_{i+1}$ is sampled from the standard normal distribution. 

```python
# random number generator 
rng = np.random.default_rng(42)
paths = 50
points = 1000
mu, sigma = 0.0, 1.0
Z = rng.normal(mu, sigma, (paths, points))
print(Z.shape)
```

```python
# assume fixed time distance for now 
interval = [0.0, 1.0]
dt = (interval[1] - interval[0]) / (points - 1)
```

```python
t_axis = np.linspace(interval[0], interval[1], points)
```

```python
W = np.zeros((paths, points))
for idx in range(points - 1):
    real_idx = idx + 1
    W[:, real_idx] = W[:, real_idx - 1] + np.sqrt(dt) * Z[:, idx]
```

```python
# Plotting 
fig, ax = plt.subplots(1, 1, figsize=(12, 8))
for path in range(paths):
    ax.plot(t_axis, W[path, :])
ax.set_title("Standard Brownian Motion sample paths")
ax.set_xlabel("Time")
ax.set_ylabel("Value")
plt.show()
```
